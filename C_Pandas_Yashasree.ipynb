```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Create a small sample DataFrame
data = {
    "Name": ["Alice", "Bob", "Charlie", "David", "Eva"],
    "Age": [24, 27, 22, 32, 29],
    "City": ["New York", "Los Angeles", "Chicago", "Houston", "Phoenix"]
}

df = pd.DataFrame(data)
df
```

```python
# Save the DataFrame to a CSV file
df.to_csv("people.csv", index=False)

# Now read it back using pandas read_csv()
df_csv = pd.read_csv("people.csv")

df_csv

```

```python
# Display the first few rows
print("First two rows:")
print(df_csv.head(2))

# Select a single column
print("\nSelecting the 'Name' column:")
print(df_csv["Name"])

# Select multiple columns
print("\nSelecting 'Name' and 'City' columns:")
print(df_csv[["Name", "City"]])
```

```python
# Using loc (label-based indexing)
print("Row with index label 2 using loc:")
print(df_csv.loc[2])

# Using iloc (integer position-based indexing)
print("\nRow at position 3 using iloc:")
print(df_csv.iloc[3])
```

```python
# Filter data: show people older than 25
print("People older than 25:")
print(df_csv[df_csv["Age"] > 25])
```

```python
# Simple data visualization: Bar chart of Ages
plt.figure(figsize=(6,4))
plt.bar(df_csv["Name"], df_csv["Age"], color="skyblue")
plt.title("Age of People by Name")
plt.xlabel("Name")
plt.ylabel("Age")
plt.show()
```
